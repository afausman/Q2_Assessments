```python
from datetime import datetime

def format_event_deadline():
    # Ask the user for the required event details
    event_title = input("Enter event title: ")
    year = int(input("Enter year: "))
    month = int(input("Enter month: "))
    day = int(input("Enter day: "))

    # Create a datetime object to process the date
    event_date = datetime(year, month, day)

    # Display the complete formatted output
    print("\nFormatted Output:")
    print(f"Event: {event_title.upper()}")
    print(f"Date: {event_date.strftime('%B')} {event_date.day}, {event_date.year}")
    print(f"Year: {year}")
    print(f"Month: {event_date.strftime('%B')}")
    print(f"Day: {day}")
    print(f"Weekday: {event_date.strftime('%A')}")

# Run the program
if __name__ == "__main__":
    format_event_deadline()
```
